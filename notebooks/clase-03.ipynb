## 1. Representando restricciones como Ax=b

Un mini-perceptrón con 2 entradas + sesgo: $\text{salida} = w_1x_1 + w_2x_2 + b$.
Tres restricciones EXACTAS: $(1,2)\to9$, $(2,0)\to5$, $(0,3)\to10$.

Cada fila de `A`: coeficiente de $w_1$, de $w_2$, del sesgo (siempre 1). `np.linalg.solve`
resuelve con eliminación de Gauss + sustitución hacia atrás por debajo — NO con Gauss-Jordan
(más barato computacionalmente, ver la Sección 9 del guion).


```python
A = np.array([[1, 2, 1],
              [2, 0, 1],
              [0, 3, 1]])
b = np.array([9, 5, 10])

w = np.linalg.solve(A, b)      # Gauss internamente
print("w = [w1, w2, b] =", w)
```

## 2. Verificación con forward pass

Un **forward pass** es evaluar el modelo hacia adelante: multiplicar las entradas originales
por los pesos encontrados y comparar con lo exigido. Si el sistema se resolvió bien, deben
coincidir EXACTO (no aproximado — a diferencia de la regresión de la Clase 02).


```python
forward = A @ w
print("forward pass =", forward)
print("salidas exigidas =", b)
print("¿coincide exacto?", np.allclose(forward, b))
```

## 3. Cuando el sistema NO tiene solución única

Si dos restricciones son (casi) la misma, el sistema queda con menos de 3 ecuaciones
independientes para 3 incognitas — `np.linalg.solve` lanza `LinAlgError`, igual que con la
matriz singular de la Clase 02.


```python
A_dep = np.array([[1, 2, 1],
                   [2, 0, 1],
                   [1, 2, 1]])   # fila 3 = fila 1 (restricción repetida, no aporta nada nuevo)
b_dep = np.array([9, 5, 9])

print("det(A_dep) =", np.linalg.det(A_dep))

try:
    np.linalg.solve(A_dep, b_dep)
except np.linalg.LinAlgError as e:
    print("⚠️ No se pudo resolver:", e)
    print("   Causa: la restricción 3 es idéntica a la 1 — solo 2 ecuaciones independientes para 3 incognitas.")
```
